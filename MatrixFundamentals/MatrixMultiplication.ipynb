# 4 Ways of Matrix * Matrix

Matrix algebra heavily utilises matrix multiplications. It is very helpful to comprehend the 4 ways of calulating a $\mathbf{A} \mathbf{B}$ matrix multiplication. The linear combination mindset is probably most helpful.

```python
from sympy import Matrix
# this package is probably not useful for real data processing,
# however here it is useful to straightforwardly apply
# inner/outer products and linear combinations
# without the 'pain' of dealing with rank 1 numpy arrays
```

```python
A = Matrix([
    [1,4],
    [2,5],
    [3,6]
])
B = Matrix([
    [7,9,11],
    [8,10,12]
])
```

```python
A
```

```python
B
```

```python
C = A @ B
C
```

# 1st Way: Linear Combination of Columns in A

```python
A @ B[:,0]  # yields 1st column in C
```

1st column in C goes as linear combination
$$
7 \cdot
\begin{bmatrix}
1\\2\\3
\end{bmatrix}
+
8 \cdot
\begin{bmatrix}
4\\5\\6
\end{bmatrix}
=
\begin{bmatrix}
39\\54\\69
\end{bmatrix}
$$

```python
A @ B[:,1]  # yields 2nd column in C
```

2nd column in C goes as linear combination
$$
9 \cdot
\begin{bmatrix}
1\\2\\3
\end{bmatrix}
+
10 \cdot
\begin{bmatrix}
4\\5\\6
\end{bmatrix}
=
\begin{bmatrix}
49\\68\\87
\end{bmatrix}
$$

```python
A @ B[:,2]  # yields 3rd column in C
```

# 2nd Way: Linear Combinations of Rows in B

```python
A[0, :] @ B  # yields 1st row in C
```

1st row in C goes as linear combination
$$
1 \cdot
\begin{bmatrix}
7 & 9 & 11
\end{bmatrix}
+
4 \cdot
\begin{bmatrix}
8 & 10 & 12
\end{bmatrix}
=
\begin{bmatrix}
39 & 49 & 59
\end{bmatrix}
$$

```python
A[1, :] @ B  # yields 2nd row in C
```

2nd row in C goes as linear combination
$$
2 \cdot
\begin{bmatrix}
7 & 9 & 11
\end{bmatrix}
+
5 \cdot
\begin{bmatrix}
8 & 10 & 12
\end{bmatrix}
=
\begin{bmatrix}
54 & 68 & 82
\end{bmatrix}
$$

```python
A[2, :] @ B  # yields 3rd row in C
```

# 3rd Way: Sum of Outer Products

```python
C1 = A[:,0] @ B[0,:]  # 1st outer product, this yields a rank-1 matrix
C1
```

1st outer product
$$
\begin{bmatrix}
1 \\ 2 \\ 3
\end{bmatrix}
\begin{bmatrix}
7 & 9 & 11
\end{bmatrix}
=
\begin{bmatrix}
7 & 9 & 11\\
14 & 18 & 22\\
21 & 27 & 33
\end{bmatrix}
$$

```python
C2 = A[:,1] @ B[1,:]  # 2nd outer product, this yields a rank-1 matrix
C2
```

```python
C1 + C2  # sum of outer products yields C, this is a rank-2 matrix
```

# 4th Way: Inner Products

We probably initially learned this in a linear algebra course, as this calculation rule is derived from theory of matrix based linear transformation...many inner products...probably most un-intuitive as it somewhat hides the essence of linear combinations/superposition of rank-1 matrices.

```python
for rowsA in range(3):
    for columnsB in range(3):
        print(A[rowsA, :] @ B[:, columnsB])
```

1st inner product, i.e. result of $C_{row 1, column 1}$
$$
\begin{bmatrix}
1 & 4
\end{bmatrix}
\begin{bmatrix}
7 \\ 8
\end{bmatrix}
=
39
$$
