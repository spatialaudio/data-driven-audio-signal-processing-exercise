# Right Inverse and its Projections

```python
import numpy as np
from numpy.linalg import matrix_rank
from scipy.linalg import svd, diagsvd, inv, pinv
```

```python
A = np.array([  # short/wide, full row rank
    [-2, -4, 0],
    [1, 4, 2]])
A
```

```python
M, N = A.shape
I_M, I_N = np.eye(M), np.eye(N)
R = matrix_rank(A)
R, R == np.min((M, N))  # rank 2 = full row rank
```

```python
# pure column space (CS), left null space (LNS) is only the zero vector
y = np.array([[6], [-3]])
y
```

```python
# pure row space (RS)
x1 = np.array([[-1], [-1], [1]])
x1
```

```python
# row space + null space (NS)
x2 = np.array([[1], [-2], [2]])
x2
```

```python
# pure null space
e_NS = x2 - x1
e_NS, A @ e_NS
```

```python
A_ri = pinv(A)  # for full row rank A this yields the right inverse
A_ri
```

```python
# make sure to understand difference between inv() and pinv()
```

```python
A.T @ inv(A @ A.T)  # explicit equation for the right inverse
```

```python
xh = A_ri @ y
np.allclose(x1, xh)  # xh is purely in row space, and thus == x1, this is the shortest vector x
# that yields A * x = y, hence ||x||_2 is min
```

```python
# projection matrices
P_CS = A @ A_ri  # to column space (CS) -> i.e. the identity matrix for short/wide, full row rank A
P_LNS = I_M - P_CS  # to left null space (LNS) -> i.e. zero matrix for short/wide, full row rank A
P_RS = A_ri @ A  # to row space (RS)
P_NS = I_N - P_RS  # to null space (NS)
P_CS, P_LNS, P_RS, P_NS
```

```python
# check projections
P_CS @ y  # projects y to itself, because P_CS = I
```

```python
P_RS @ x1, P_NS @ x1  # x1: pure row space split to RS and NS
```

```python
P_RS @ x2, P_NS @ x2  # x2: row space + null space split to RS and NS
```

```python
np.allclose(np.array([[0], [0], [0]]), P_NS @ x1), np.allclose(e_NS, P_NS @ x2)
```

```python
# pseudo-inverse via SVD
[U, s, Vt] = svd(A)
V = Vt.T
S = diagsvd(s, M, N)
A_ri_svd = V @ diagsvd(1/s, N, M) @ U.T  # general concept of the pseudo inverse
A_ri_svd, np.allclose(A_ri, A_ri_svd)
```

```python
# special case: short/wide, full row rank == right inverse
V @ (S.T @ inv(S @ S.T)) @ U.T  # formulation for the right inverse using the S matrix
```

```python
# general formulation for the projections matrices based on SVD matrices
U[:,:R] @ U[:,:R].T  # P_CS, i.e. the identity matrix for short/wide, full row rank A
```

```python
U[:,R:] @ U[:,R:].T  # P_LNS, i.e. zero matrix for short/wide, full row rank A
```

```python
V[:, :R] @ V[:, :R].T  # P_RS
```

```python
V[:, R:] @ V[:, R:].T  # P_NS
```
