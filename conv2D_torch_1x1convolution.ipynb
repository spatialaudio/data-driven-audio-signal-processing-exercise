# Exercise: 1x1 Convolution

- we create output feature maps with same dimension of input feature maps by so called 1x1 convolution
- this allows feature dimensions to be squeezed (which is very often used) or even blown up but retaining the image size
- one 1x1 convolution goes by: a weighted superposition of input feature map pixels at the very same position is performed to yield one output feature map pixel at the very same pixel position
- below we demonstrate this with some random integer tensors using torch's Conv2d() with `kernel_size=1`
- recall that in ML **convolutional layer** typically calculate the **correlation**
- in our case we calculate inner products, so we don't need to care to much about the `conv` vs. `corr` issue

```python
import numpy as np
import torch
rng = np.random.default_rng(1)
```

```python
H, W = 32, 16  # image size in pixels
in_channels = 8
out_channels = 4
```

```python
f = torch.nn.Conv2d(in_channels=in_channels,
                    out_channels=out_channels,
                    kernel_size=1,
                    stride=1, 
                    padding = 0,
                    bias=False)
# init with some nice integer-like from an outer product
tmp = (np.arange(out_channels)+1)[:, None] @ (np.arange(in_channels)+1)[None, :]
tmp = tmp[:,:,None,None]
f.weight = torch.nn.Parameter(torch.Tensor(tmp))  # torch uses float numbers
f.weight.shape, tmp.shape, tmp[:,:,0,0], f.weight[:,:,0,0]
```

```python
X = rng.integers(-3, +3, (1, in_channels, H, W))
Xt = torch.Tensor(X)
X.shape, Xt.shape
```

calc 1x1 Conv by torch

```python
Yt = f(Xt)
Yt.shape, Yt[0, 0, 0, -1], Yt[0, out_channels-1, -1, 0], Yt[0, out_channels-2, H-1, W-1]
```

check some output 'pixels' manually

```python
(Xt[0, :, 0, -1] * f.weight[0, :, 0, 0]).sum()
```

```python
(Xt[0, :, -1, 0] * f.weight[out_channels-1, :, 0, 0]).sum()
```

```python
(Xt[0, :, H-1, W-1] * f.weight[out_channels-2, :, 0, 0]).sum()
```
